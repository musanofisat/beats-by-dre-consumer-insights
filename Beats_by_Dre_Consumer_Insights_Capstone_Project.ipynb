### 4.3 Data Cleaning
Outline the steps taken to clean and preprocess the data, highlighting any significant decisions made during this process.

Load the data below and process cleaning here.

```python

# loading of the dataframe
df = pd.read_csv('/content/merged_data.csv')
df

#to
df.info()

#to check for duplicates
df.duplicated().sum(axis=0)
df = df.drop_duplicates(keep='first')

#to check for null values
df.isnull().sum()

# to split the timestamp column into two different columns
df['timestamp'] = df['timestamp'].str.replace('Reviewed in the ', '')


df['country_product_was_reviewed'] = df['timestamp'].str.extract(r'(.+?) (?:January|February|March|April|May|June|July|August|September|October|November|December)')


df['date_of_review'] = df['timestamp'].str.extract(r'((?:January|February|March|April|May|June|July|August|September|October|November|December) \d{1,2}, \d{4})')

df['date_of_review'] = pd.to_datetime(df['date_of_review'])

# to extract only the year from the date of review
df['year'] = df['date_of_review'].dt.year

#product mapping
product_mapping = {

    'B0D4SX9RC6': 'Beat pill',
    'B08X4YMTPM': 'JBL charge 5',
    'B0BX9B8V5F': 'DOSS Extreme Boom+',
    'B07PMBCTSY': 'Soundcore by anker',
    'B099TJGJ91': 'Bose SoundLink Flex',
    'B09NWBZ7T3': 'Bang & Olufsen Beosound',
    'B0CRYXN3S9': 'W-KING 300W',
    'B00K2RQ94C': 'Sennheiser Enterprise ',
    'B09XXW54QG': 'Marshall Emberton II',
    'B0CY6S748H': 'Roam 2 by Sonos',
    'B0CXKSCY1G': 'sony ULT field 1'

}
df['product_name'] = df['product_id'].map(product_mapping)

#color mappiy
color_mapping = {
    'Color: Black'        : 'Black',
    'Color: White'      : 'white',
    'Color: Gray'       : 'Grey',
    'Color: Camouflage'  : 'Green',
    'Color: Teal'       : 'Green',
    'Color: Blue'       : 'Blue',
    'Color: Red'        : 'Red',
    'Color: Green'      : 'Green',
    'Color: Pink'       : 'Pink',
    'Color: Off White'   : 'Grey',
    'Color: Orange'     : 'Orange',
    'Color: Forest Gray': 'Grey',
    'Color: Gold'       : 'Gold',
    'Color: Purple'     : 'Purple',
    'Color: Grey'       : 'Grey',
    'Color: NavyStyle: Bluetooth speaker' : 'Blue',
    'Color: Grey MistStyle: Bluetooth speaker' : 'Grey',
    'Color: Black AnthraciteStyle: Bluetooth speaker' : 'Black',
    'Color: GreenStyle: Bluetooth speaker' : 'Green',
    'Color: ChestnutStyle: Bluetooth speaker' : 'Brown',
    'Color: Stone Blue' : 'Blue',
    'Color: White Smoke' : 'Grey',
    'Color: Chilled Lilac' : 'Purple',
    'Color: Black & BrassStyle: Emberton II' : 'Black',
    'Color: BlackStyle: Emberton' : 'Black',
    'Color: CreamStyle: Emberton II' : 'Black',
    'Color: Matte BlackStyle: PillPattern: Without AppleCare+' : 'Black',
    'Color: Champagne GoldStyle: PillPattern: Without 1' : 'Gold',
    'Color: Statement RedStyle: PillPattern: Without AppleCare+' : ' Red'

}
df['color'] = df['product_attributes'].map(color_mapping)

# to drop unnecessary columns
df = df.drop(['timestamp', 'product_attributes'], axis=1)

# to fill the null values with mode
color_mode = df['color'].mode()[0]
df['color'].fillna(color_mode, inplace=True)

#to drop missing column
df.dropna(subset=['content'], inplace=True)
```

Data cleaning process

*   The merged datasets was imported into google colab for cleaning

*   Null values were checked and replaced with mode for categorical columns i.e. products attributes
*   The timestamp column consist of the country and the date was split into two columns and the year column was extracted.

## 6. Sentiment Analysis

### 6.1 Methodology
Explain the sentiment analysis techniques used (e.g., TextBlob, NLTK) and provide code snippets.

```python

import nltk
import pandas as pd
from nltk.sentiment import SentimentIntensityAnalyzer

nltk.download('vader_lexicon')

# Sample data (replace with your data)


sia = SentimentIntensityAnalyzer()
df['sentiment_score'] = df['content'].apply(lambda review: sia.polarity_scores(review)['compound'])

# Group by product and calculate average sentiment score
product_sentiments = df.groupby('product_name')['sentiment_score'].mean()
print(product_sentiments)

def categorize_sentiment(score):
    if score >= 0.05:
        return 'Positive'
    elif score <= -0.05:
        return 'Negative'
    else:
        return 'Neutral'

df['sentiment_category'] = df['sentiment_score'].apply(categorize_sentiment)

print(df)
```

### 6.2 Results
Present the sentiment analysis results, including visualizations of sentiment distribution, and discuss the overall sentiment towards Beats by Dre products.


```python

# Pivot data
pivot_data = df.pivot_table(index='product_name', columns='sentiment_category',
                            aggfunc='size', fill_value=0)

# Create stacked bar chart
pivot_data.plot(kind='barh', stacked=True, figsize=(12, 12))

# Add labels and title
plt.title('Sentiment Analysis by Product', size=20)
plt.xlabel('Product Name')
plt.ylabel('Number of Reviews')
plt.legend(title='Sentiment')
plt.xticks(rotation=90)
plt.show()
```

According to the sentiment result, Beat pill has a more positive sentiment against the neutral and negative sentiment which is a good result for the beat products.

```python
import os

import google.generativeai as genai

# Load your dataset
df = pd.read_csv('/content/beats_reviews.csv')

# Replace 'your_api_key' with the actual API key
api_key = os.environ.get("GEMINI_API_KEY")  # Set this securely; do not hard-code API keys

# Function to analyze reviews using Gemini AI
def analyze_reviews(reviews, api_key,prompt_message):
    genai.configure(api_key=api_key)

    model = genai.GenerativeModel(
        model_name="gemini-1.5-flash",
        generation_config={
            "temperature": 1,
            "top_p": 0.95,
            "top_k": 64,
            "max_output_tokens": 8192,
            "response_mime_type": "text/plain",
        },
    )

    chat_session = model.start_chat(history=[])

    prompt = prompt_message + "\n\n".join(reviews)
    response = chat_session.send_message(prompt)
    return response.text
```

### 8.2 Comparison of Sentiments
Compare the sentiment analysis results of Beats by Dre with its competitors using visualizations to highlight differences and similarities.

```python
from textblob import TextBlob
# Perform sentiment analysis using TextBlob
df['sentiment'] = df['content'].apply(lambda x: TextBlob(x).sentiment.polarity)

# Classify sentiment into positive, negative, and neutral
df['sentiment_category'] = pd.cut(df['sentiment'],  bins=[-1, -0.01, 0.01, 1],
                                  labels=['Negative', 'Neutral', 'Positive'],
                               include_lowest=True)

# Count sentiment categories
sentiment_counts = df['sentiment_category'].value_counts()

# Bar Chart
plt.figure(figsize=(12, 10))
plt.bar(sentiment_counts.index, sentiment_counts.values)
plt.xlabel('Sentiment')
plt.ylabel('Count of reviews')
plt.title('Sentiment Analysis for textblob', size = 20)
plt.show()
# Pivot data
pivot_data = df.pivot_table(index='product_name', columns='sentiment_category',
                            aggfunc='size', fill_value=0)

# Create stacked bar chart
pivot_data.plot(kind='barh', stacked=True, figsize=(12, 12))

# Add labels and title
plt.title('Sentiment Analysis by Product', size=20)
plt.xlabel('Product Name')
plt.ylabel('Number of Reviews')
plt.legend(title='Sentiment')
plt.xticks(rotation=90)
plt.show()
```

Observation: using the textblob, the positive reviews made a significant dominance over the negative and neutral reviews across all products. JBL, Bose and W-king do not have any negative reviews. Sonos and Anker has the most negative reviews.
